```python
print("\n===== Importing Required Libraries =====\n")
# ================================================================
# 2. Importing Required Libraries
# ================================================================

# ================================================================
# 2.1 Core Python & Utility Libraries
# ================================================================
import os
import re
import time
import random
import warnings
warnings.filterwarnings('ignore')

# ================================================================
# 2.2 Data Handling & Processing Libraries
# ================================================================
import pandas as pd
import numpy as np

# ================================================================
# 2.3 Progress Monitoring Libraries
# ================================================================
from tqdm import tqdm

# ================================================================
# 2.4 Data Visualization (EDA + Results)
# ================================================================
import matplotlib.pyplot as plt
import seaborn as sns

# ================================================================
# 2.5 LLM / Model APIs
# ================================================================
from groq import Groq
from openai import OpenAI

# ================================================================
# 2.6 Evaluation Metrics
# ================================================================
from bert_score import score as bertscore
from rouge_score import rouge_scorer
from sklearn.metrics import accuracy_score

# ================================================================
# 2.7 Explainable AI
# ================================================================
from lime.lime_text import LimeTextExplainer
import shap
import lime
from captum.attr import IntegratedGradients

# ================================================================
# 2.8 LIBRARY VERSIONS
# ================================================================

print("===== LIBRARY VERSIONS =====\n")

from importlib.metadata import version, PackageNotFoundError

def get_version(pkg_name):
    try:
        return version(pkg_name)
    except PackageNotFoundError:
        return "Not installed"
    except Exception:
        return "Version not found"

# Core Python
import platform
print("Python:", platform.python_version())

# Data
print("pandas:", pd.__version__)
print("numpy:", np.__version__)

# Progress
import tqdm

print("tqdm:", tqdm.__version__)

# Visualization
import matplotlib
print("matplotlib:", matplotlib.__version__)
print("seaborn:", sns.__version__)

# LLM APIs
print("groq:", get_version("groq"))
print("openai:", get_version("openai"))

# Evaluation
print("rouge-score:", get_version("rouge-score"))
print("scikit-learn:", get_version("scikit-learn"))

# Explainability
print("lime:", get_version("lime"))
print("shap:", get_version("shap"))

# Built-in
print("re: built-in (no version)")

# ================================================================
# 3. GROQ API key Loading
# ================================================================
print("\n===== GROQ API key Loading =====\n")

from google.colab import userdata

GROQ_KEYS = [
    userdata.get('GROQ_API_KEY1'),
    userdata.get('GROQ_API_KEY2'),
    userdata.get('GROQ_API_KEY3'),
    userdata.get('GROQ_API_KEY4'),
    userdata.get('GROQ_API_KEY5'),
    userdata.get('GROQ_API_KEY6'),
    userdata.get('GROQ_API_KEY7'),
    userdata.get('GROQ_API_KEY8'),
    userdata.get('GROQ_API_KEY9'),
    userdata.get('GROQ_API_KEY10'),
    userdata.get('GROQ_API_KEY11'),
    userdata.get('GROQ_API_KEY12')
]

print("\n===== GROQ API keys Loaded Successfully from Colab Secrets =====\n")


print("\n===== LOADING COMMON BFSI DATASET =====\n")

# ================================================================
# 4. LOAD PREPROCESSED BFSI DATASET
# ================================================================

# The dataset used in this experiment was preprocessed
# using a separate dataset preparation pipeline.
# The cleaned and validated dataset is loaded directly
# for prompt engineering experiments.

# ================================================================
# LOAD COMMON DATASET
# ================================================================

df = pd.read_csv(

    "/content/common_bfsi_dataset.csv"
)

print(
    "Dataset loaded successfully"
)

print(
    "Dataset Shape:",
    df.shape
)

# ================================================================
# DISPLAY SAMPLE RECORDS
# ================================================================

print("\n===== SAMPLE RECORDS =====\n")

print(
    df[[
        'Product',
        'Issue',
        'clean_text'
    ]].head(3)
)

# ================================================================
# DATASET VALIDATION
# ================================================================

print("\n===== DATASET VALIDATION =====\n")

print(
    "Null values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Columns:",
    list(df.columns)
)

print("\n===== FINAL EXPERIMENT DATASET =====\n")

print(
    "Dataset Shape:",
    df.shape
)

print(
    "Total Records:",
    len(df)
)


# ================================================================
# 5. Model and Task
# ================================================================

print("\n===== MODEL & TASK =====\n")
print("\n=====   MODEL:  openai/gpt-oss-120b  =====\n")
print("\n=====   TASK2:  Customer Communication Drafting   =====\n")

MODEL_NAME = "openai/gpt-oss-120b"
MODEL_SHORT = "gptoss120b"
TASK_NAME = "Task-2: Customer Communication Drafting"

MODELS = [
    "openai/gpt-oss-120b"
]

current_key_index = 0
blocked_keys = {}
error_counter = 0

def run_model(prompt, model_name, retries=5):

    global current_key_index
    global error_counter

    for attempt in range(retries):

        try:

            start_index = current_key_index

            while True:

                key = GROQ_KEYS[current_key_index]

                if key not in blocked_keys or time.time() > blocked_keys[key]:
                    break

                current_key_index = (
                    current_key_index + 1
                ) % len(GROQ_KEYS)

                if current_key_index == start_index:

                    time.sleep(60)

            client = Groq(api_key=key)

            current_key_index = (
                current_key_index + 1
            ) % len(GROQ_KEYS)

            time.sleep(random.uniform(1, 3))

            response = client.chat.completions.create(

                model=model_name,

                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],

                temperature=0.3,

                top_p=0.9,

                max_tokens=300,

                timeout=60
            )

            output = response.choices[0].message.content

            if output is None:
                return "ERROR"

            return output

        except Exception as e:

            error_counter += 1

            blocked_keys[key] = time.time() + 120

            time.sleep(5)

    return "ERROR"


# ================================================================
# 6. Generic Experiment Engine
# ================================================================

def run_experiment(df, prompts):

    results = []

    for i, row in tqdm.tqdm(
        df.iterrows(),
        total=len(df)
    ):

        text = row['clean_text']

        for p_type, func in prompts.items():

            prompt = func(text)

            output = run_model(
                prompt,
                MODEL_NAME
            )

            results.append({

                "id": i,

                "model": MODEL_NAME,

                "prompt_type": p_type,

                "input_text": text,

                "output": output
            })

        # ========================================================
        # CHECKPOINT SAVE EVERY 25 RECORDS
        # ========================================================

        if i % 25 == 0 and i != 0:

            checkpoint_df = pd.DataFrame(results)

            checkpoint_df.to_excel(

                f"{TASK_NAME}_{MODEL_SHORT}_checkpoint.xlsx",

                index=False
            )

    return pd.DataFrame(results)

# ================================================================
# 7. PROMPTS
# ================================================================


# ------------------------------------------------
# 7.1 ZERO SHOT
# ------------------------------------------------

def zero_shot_prompt(text):

    return f"""

Draft a professional customer service response
for the following banking complaint.

Complaint:
{text}

Response:

"""


# ------------------------------------------------
# 7.2 FEW SHOT
# ------------------------------------------------

def few_shot_prompt(text):

    return f"""

Example:

Complaint:
The customer was charged unauthorized fees.

Response:
We sincerely apologize for the inconvenience caused.
Our team will investigate the unauthorized charges
and provide an update shortly.

Now respond to this complaint.

Complaint:
{text}

Response:

"""


# ------------------------------------------------
# 7.3 ROLE BASED
# ------------------------------------------------

def role_based_prompt(text):

    return f"""

You are a senior BFSI customer support manager.

Draft a professional, empathetic, and compliant
response to the customer complaint.

Complaint:
{text}

Response:

"""


# ------------------------------------------------
# 7.4 INSTRUCTION PROMPT
# ------------------------------------------------

def instruction_prompt(text):

    return f"""

Draft a banking customer response.

Instructions:

1. Be professional
2. Show empathy
3. Acknowledge the issue
4. Avoid legal admissions
5. Provide resolution direction

Complaint:
{text}

Response:

"""


# ------------------------------------------------
# 7.5 CONSTRAINT PROMPT
# ------------------------------------------------

def constraint_prompt(text):

    return f"""

Respond ONLY using information present
in the complaint.

Do NOT assume facts.
Do NOT promise unavailable resolutions.

Complaint:
{text}

Response:

"""


# ------------------------------------------------
# 7.6 CHAIN OF THOUGHT (CoT)
# ------------------------------------------------

def cot_prompt(text):

    return f"""

You are a BFSI customer communication expert.

Step 1:
Understand the customer issue.

Step 2:
Identify customer concerns.

Step 3:
Determine appropriate tone.

Step 4:
Draft a professional response.

Complaint:
{text}

Response:

"""

task2_prompts = {
     "zero_shot": zero_shot_prompt,
    "few_shot": few_shot_prompt,
    "role_based": role_based_prompt,
    "instruction": instruction_prompt,
    "constraint": constraint_prompt,
     "cot": cot_prompt
}

print("\n===== TASK2:  Customer Communication Drafting - EXPERIMENT EXECUTION =====\n")

# ================================================================
# 8. EXECUTION
# ================================================================

task2_df = run_experiment(df, task2_prompts)

# ================================================================
# 9. METRICS
# ================================================================

# ================================================================
# 9.1 ROUGE
# ================================================================

from rouge_score import rouge_scorer

scorer = rouge_scorer.RougeScorer(
    ['rouge1', 'rougeL'],
    use_stemmer=True
)

def rouge_l(output, reference):

    return scorer.score(
        reference,
        output
    )['rougeL'].fmeasure


# ================================================================
# 9.2 POLITENESS SCORE
# ================================================================

def politeness_score(text):

    keywords = [

        "sorry",

        "regret",

        "thank",

        "appreciate",

        "apologize"
    ]

    return sum(

        k in text.lower()

        for k in keywords

    ) / len(keywords)


# ================================================================
# 9.3 PROFESSIONALISM SCORE
# ================================================================

def professionalism_score(text):

    keywords = [

        "customer",

        "review",

        "investigation",

        "complaint",

        "support"
    ]

    return sum(

        k in text.lower()

        for k in keywords

    ) / len(keywords)

# ================================================================
# 9.4 COMPLIANCE RISK
# ================================================================

def compliance_risk(text):

    text = str(text).lower()

    compliance_indicators = [

        "policy",

        "regulation",

        "compliance",

        "review",

        "investigate",

        "resolve",

        "guidelines",

        "process",

        "verification",

        "authorized",

        "security",

        "privacy",

        "risk",

        "support",

        "documentation",

        "customer",

        "account",

        "official",

        "bank",

        "financial"
    ]

    matches = sum(

        term in text

        for term in compliance_indicators
    )

    score = matches / len(compliance_indicators)

    return min(score, 1)

# ================================================================
# 9.5 CLARITY SCORE
# ================================================================

def clarity_score(text):

    words = len(

        str(text).split()
    )

    return min(words / 100, 1)


# ================================================================
# 9.6 HALLUCINATION SCORE
# ================================================================

def communication_hallucination(text):

    text = str(text).lower()

    risky_terms = [

        # overconfidence
        "guarantee",
        "certainly",
        "definitely",
        "always",
        "never fail",
        "100%",

        # unsupported approval/verification
        "approved",
        "confirmed",
        "verified",
        "authorized",

        # legal/compliance risk
        "legally",
        "compliant",
        "risk-free",

        # unrealistic promises
        "immediately resolved",
        "permanently fixed",
        "fully secure",

        # unsupported assumptions
        "assume",
        "no risk",
        "without issue"
    ]

    matches = sum(

        term in text

        for term in risky_terms
    )

    score = matches / len(risky_terms)

    return min(score, 1)


print("\n===== OUTPUT CLEANING =====\n")

# ================================================================
# REMOVE EMPTY / ERROR OUTPUTS
# ================================================================

task2_df = task2_df[
    task2_df['output'].notna()
]

task2_df = task2_df[
    task2_df['output'].str.strip() != ""
]

task2_df = task2_df[

    ~task2_df['output'].str.startswith(
        "ERROR"
    )
]

task2_df = task2_df.reset_index(drop=True)

print("Remaining rows:", len(task2_df))


print("\n===== METRIC CALCULATION =====\n")

# ================================================================
# 10. APPLY METRICS
# ================================================================

task2_df['politeness'] = task2_df[
    'output'
].apply(politeness_score)

task2_df['professionalism'] = task2_df[
    'output'
].apply(professionalism_score)

task2_df['compliance_risk'] = task2_df[
    'output'
].apply(compliance_risk)

task2_df['clarity'] = task2_df[
    'output'
].apply(clarity_score)

task2_df['hallucination'] = task2_df[
    'output'
].apply(communication_hallucination)


# ================================================================
# 11. ROUGE-L
# ================================================================

# ================================================================
# 11.1 CREATE SIMPLE REFERENCE SUMMARY
# ================================================================

def create_reference_summary(text):

    text = str(text)

    return text[:300]


task2_df['reference_summary'] = task2_df[
    'input_text'
].apply(create_reference_summary)


# ================================================================
# 11.2 ROUGE-L USING REFERENCE SUMMARY
# ================================================================

task2_df['rouge_l'] = task2_df.apply(

    lambda x: rouge_l(

        x['output'],

        x['reference_summary']

    ),

    axis=1
)


# ================================================================
# 12. TASK 2 SUMMARY
# ================================================================

summary_task2 = task2_df.groupby(

    ['prompt_type']

).agg({

    'politeness': 'mean',

    'professionalism': 'mean',

    'compliance_risk': 'mean',

    'clarity': 'mean',

    'hallucination': 'mean',

    'rouge_l': 'mean'

}).reset_index()

print("\n===== TASK 2 SUMMARY =====\n")

display(summary_task2)

# ================================================================
# 13. VISUALIZATION
# ================================================================

# ================================================================
# COLOR PALETTE
# ================================================================

prompt_palette = {

    'zero_shot': '#1f77b4',
    'few_shot': '#ff7f0e',
    'role_based': '#2ca02c',
    'instruction': '#d62728',
    'constraint': '#9467bd',
    'cot': '#8c564b'
}

plt.figure(figsize=(10, 5))

sns.barplot(

    data=summary_task2,

    x='prompt_type',

    y='politeness'
)

plt.title(

    "Task 2 - Politeness Score"
)

plt.xticks(rotation=45)

plt.show()


# ================================================================
# 14. BERTScore Evaluation
# ================================================================

print("\n===== BERTScore EVALUATION =====\n")


P, R, F1 = bertscore(

    task2_df['output'].astype(str).tolist(),

    task2_df['reference_summary'].astype(str).tolist(),

    lang='en',

    model_type='roberta-large',

    verbose=True
)


# ================================================================
# 14.1 STORE SCORES
# ================================================================

task2_df['bert_precision'] = P.numpy()

task2_df['bert_recall'] = R.numpy()

task2_df['bert_f1'] = F1.numpy()


# ================================================================
# 14.2 SUMMARY
# ================================================================

bert_summary = task2_df.groupby(

    ['prompt_type']

).agg({

    'bert_f1': 'mean'

}).reset_index()


print("\n===== BERTScore Summary =====\n")

display(bert_summary)

# ================================================================
# 15. FINAL TASK 2 SUMMARY
# ================================================================

final_task2_summary = task2_df.groupby(

    ['model', 'prompt_type']

).agg({

    'politeness': 'mean',

    'professionalism': 'mean',

    'compliance_risk': 'mean',

    'clarity': 'mean',

    'hallucination': 'mean',

    'rouge_l': 'mean',

    'bert_f1': 'mean'

}).reset_index()


print(
    "\n===== FINAL TASK 2 SUMMARY =====\n"
)

display(final_task2_summary)


print("\n===== PERFORMANCE HEATMAP =====\n")

# ================================================================
# PERFORMANCE HEATMAP
# ================================================================

heatmap_data = final_task2_summary.set_index(

    'prompt_type'

)[[

    'politeness',

    'professionalism',

    'compliance_risk',

    'clarity',

    'hallucination',

    'rouge_l',

    'bert_f1'
]]

plt.figure(figsize=(12,6))

sns.heatmap(

    heatmap_data,

    annot=True,

    cmap='Blues'
)

plt.title("TASK-2:  Customer Communication Drafting - Prompting Performance Heatmap")


plt.show()

print("\n===== SAMPLE OUTPUT ANALYSIS =====\n")

# ================================================================
# 16. SAMPLE OUTPUT ANALYSIS
# ================================================================

for p in task2_df['prompt_type'].unique():

    print(f"\n\n===== TASK2:  Customer Communication Drafting : {p.upper()} =====\n")

    sample_output = task2_df[

        task2_df['prompt_type'] == p

    ]['output'].iloc[0]

    print(sample_output[:1500])

    print("\n" + "="*80)

print("\n===== LIME EXPLAINABILITY =====\n")

# ================================================================
# 17. LIME EXPLAINABILITY
# ================================================================

print("Running LIME Explainability...")


# ================================================================
# 17.1 LIME EXPLAINER
# ================================================================

explainer = LimeTextExplainer(

    class_names=['Neutral', 'Customer_Response']
)


# ================================================================
# 17.2 HEURISTIC PREDICTOR
# ================================================================


def predictor(texts):

    response_words = [

        "customer",
        "complaint",
        "response",
        "review",
        "support",
        "assistance",
        "investigation",
        "verification",
        "documentation",
        "records",
        "compliance",
        "resolution",
        "communication",
        "account",
        "credit",
        "identity",
        "fraud",
        "equifax",
        "proof",
        "evidence"
    ]

    probs = []

    for text in texts:

        text = str(text).lower()

        score = 0

        for word in response_words:

            if word in text:

                score += 1

        probability = 1 / (

            1 + np.exp(-score)

        )

        probs.append([

            1 - probability,

            probability

        ])

    return np.array(probs)


# ================================================================
# 17.3 SELECT REPRESENTATIVE GENERATED OUTPUT
# ================================================================

if 'cot' in task2_df['prompt_type'].values:

    sample_rows = task2_df[

        task2_df['prompt_type'] == 'cot'
    ]

else:

    sample_rows = task2_df


sample_rows = sample_rows[

    sample_rows['output'].notna()
]

sample_rows = sample_rows[

    sample_rows['output'].str.strip() != ""
]

sample_rows = sample_rows[

    sample_rows['output'] != "ERROR"
]


sample_text = sample_rows['output'].iloc[0]


# ================================================================
# 17.4 GENERATE EXPLANATION
# ================================================================

exp = explainer.explain_instance(

    sample_text,

    predictor,

    num_features=10
)

# ================================================================
# 17.5 FEATURE IMPORTANCE TABLE
# ================================================================

lime_features = exp.as_list()

lime_df = pd.DataFrame(

    lime_features,

    columns=['token', 'importance']
)

lime_df = lime_df.sort_values(

    by='importance',

    ascending=False
)

display(lime_df)

# ================================================================
# 18. SHAP EXPLAINABILITY
# ================================================================

print("\n===== SHAP EXPLAINABILITY =====\n")


# ================================================================
# 18.1 SHAP PREDICTOR
# ================================================================

# ================================================================
# TASK 2 SHAP PREDICTOR
# ================================================================

def shap_predict(texts):

    response_words = [

        "customer",
        "complaint",
        "response",
        "review",
        "support",
        "assistance",
        "investigation",
        "verification",
        "documentation",
        "records",
        "compliance",
        "resolution",
        "communication",
        "account",
        "credit",
        "identity",
        "fraud",
        "equifax",
        "proof",
        "evidence"
    ]

    scores = []

    for text in texts:

        text = str(text).lower()

        score = 0

        for word in response_words:

            if word in text:

                score += 1

        probability = 1 / (

            1 + np.exp(-score)

        )

        scores.append(probability)

    return np.array(scores)

# ================================================================
# 18.2 SHAP EXPLAINER
# ================================================================

explainer = shap.Explainer(

    shap_predict,

    shap.maskers.Text()
)


# ================================================================
# 18.3 SELECT REPRESENTATIVE GENERATED OUTPUT
# ================================================================

cot_rows = task2_df[

    task2_df['prompt_type'] == 'cot'
]

if len(cot_rows) > 0:

    sample_text = cot_rows[
        'output'
    ].iloc[0]

else:

    sample_text = task2_df[
        'output'
    ].iloc[0]


# ================================================================
# 18.4 GENERATE SHAP VALUES
# ================================================================

shap_values = explainer([sample_text])

# ================================================================
# 18.5 FEATURE IMPORTANCE TABLE
# ================================================================

shap_tokens = shap_values[0].data

shap_scores = shap_values[0].values


shap_df = pd.DataFrame({

    'token': shap_tokens,

    'importance': shap_scores
})

# ================================================================
# CLEAN SHAP TOKENS
# ================================================================

shap_df = shap_df[

    shap_df['token'].str.len() > 2
]

shap_df = shap_df[

    ~shap_df['token'].str.contains(
        r'[^a-zA-Z]'
    )
]

# ================================================================
# 18.6 ABSOLUTE IMPORTANCE
# ================================================================

shap_df['abs_importance'] = (

    shap_df['importance'].abs()
)

shap_df = shap_df.sort_values(

    by='abs_importance',

    ascending=False
)

shap_df['abs_importance'] = (

    shap_df['importance'].abs()
)

shap_df = shap_df.sort_values(

    by='abs_importance',

    ascending=False
)

display(shap_df.head(15))



# ================================================================
# 19. INTEGRATED GRADIENTS (GENERIC TRANSFORMER VERSION)
# ================================================================

import torch

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

from captum.attr import IntegratedGradients


print("\n===== INTEGRATED GRADIENTS EXPLAINABILITY =====\n")

# ================================================================
# 19.1 LOAD GENERIC DISTILBERT
# ================================================================

tokenizer = AutoTokenizer.from_pretrained(

    "distilbert-base-uncased-finetuned-sst-2-english"
)

model = AutoModelForSequenceClassification.from_pretrained(

    "distilbert-base-uncased-finetuned-sst-2-english"
)

model.eval()


# ================================================================
# 19.2 REPRESENTATIVE GENERATED OUTPUT
# ================================================================

cot_rows = task2_df[

    task2_df['prompt_type'] == 'cot'
]

if len(cot_rows) > 0:

    sample_text = cot_rows[
        'output'
    ].iloc[0]

else:

    sample_text = task2_df[
        'output'
    ].iloc[0]


# ================================================================
# 19.3 TOKENIZATION
# ================================================================

inputs = tokenizer(

    sample_text,

    return_tensors="pt",

    truncation=True,

    padding=True
)


# ================================================================
# 19.4 EMBEDDING LAYER
# ================================================================

embedding_layer = (
    model.distilbert.embeddings.word_embeddings
)


# ================================================================
# 19.5 FORWARD FUNCTION
# ================================================================

def forward_func_emb(

    inputs_embeds,

    attention_mask
):

    outputs = model(

        inputs_embeds=inputs_embeds,

        attention_mask=attention_mask
    )

    return outputs.logits[:, 1]


# ================================================================
# 19.6 INTEGRATED GRADIENTS
# ================================================================

ig = IntegratedGradients(
    forward_func_emb
)


# ================================================================
# 19.7 INPUT EMBEDDINGS
# ================================================================

input_embeddings = embedding_layer(
    inputs['input_ids']
)


# ================================================================
# 19.8 CLEAR GRADIENTS
# ================================================================

model.zero_grad()


# ================================================================
# 19.9 COMPUTE ATTRIBUTIONS
# ================================================================

attributions, delta = ig.attribute(

    inputs=input_embeddings,

    additional_forward_args=(

        inputs['attention_mask'],
    ),

    return_convergence_delta=True
)


# ================================================================
# 19.10 TOKENS + SCORES
# ================================================================

tokens = tokenizer.convert_ids_to_tokens(
    inputs['input_ids'][0]
)

scores = attributions.sum(

    dim=-1

).squeeze(0).detach().numpy()


# ================================================================
# 19.11 DATAFRAME
# ================================================================

ig_df = pd.DataFrame({

    'token': tokens,

    'importance': scores
})


# ================================================================
# 19.12 REMOVE SPECIAL TOKENS
# ================================================================

ig_df = ig_df[

    ~ig_df['token'].isin([

        '[CLS]',

        '[SEP]'
    ])
]

# ================================================================
# 19.13 REMOVE SYMBOLS / SHORT TOKENS
# ================================================================

ig_df = ig_df[

    ig_df['token'].str.len() > 2
]

ig_df = ig_df[

    ~ig_df['token'].str.contains(
        r'[^a-zA-Z]'
    )
]


# ================================================================
# 19.14 ABSOLUTE IMPORTANCE
# ================================================================

ig_df['abs_importance'] = (

    ig_df['importance'].abs()
)


# ================================================================
# 19.15 SORT
# ================================================================

ig_df = ig_df.sort_values(

    by='abs_importance',

    ascending=False
)


# ================================================================
# 19.16 DISPLAY
# ================================================================

display(ig_df.head(20))


# ================================================================
# 19.17 VISUALIZATION
# ================================================================

plt.figure(figsize=(10,5))

top_ig = ig_df.head(10)

plt.bar(

    top_ig['token'],

    top_ig['abs_importance']
)

plt.xticks(rotation=45)

plt.title(

    "Integrated Gradients Token Importance"
)

plt.show()



# ================================================================
# 20. FINAL LIME VISUALIZATION
# ================================================================


lime_plot_df_task2 = lime_df.sort_values(
    by='importance',
    ascending=False
).head(10)

lime_colors = plt.cm.tab20(

    np.linspace(
        0,
        1,
        len(lime_plot_df_task2)
    )
)

plt.figure(figsize=(10,5))

plt.bar(

    lime_plot_df_task2['token'],

    lime_plot_df_task2['importance'],

    color=lime_colors

)

plt.xticks(rotation=45)

plt.title(
    "Task 2 - LIME Token Importance"
)

plt.show()



# ================================================================
# 21. FINAL SHAP VISUALIZATION
# ================================================================

top_shap_task2 = shap_df.sort_values(
    by='abs_importance',

    ascending=False

).head(10)

shap_colors = plt.cm.Set3(

    np.linspace(
        0,
        1,
        len(top_shap_task2)
    )
)

plt.figure(figsize=(10,5))

plt.bar(

    top_shap_task2['token'],

    top_shap_task2['abs_importance'],

    color=shap_colors

)

plt.xticks(rotation=45)

plt.title(
    "Task 2 - SHAP Token Importance"
)

plt.show()



# ================================================================
# 22. FINAL INTEGRATED GRADIENTS VISUALIZATION
# ================================================================

top_ig_task2 = ig_df.sort_values(
    by='abs_importance',

    ascending=False

).head(10)

ig_colors = plt.cm.Paired(

    np.linspace(
        0,
        1,
        len(top_ig_task2)
    )
)

plt.figure(figsize=(10,5))

plt.bar(

    top_ig_task2['token'],

    top_ig_task2['abs_importance'],

    color=ig_colors

)

plt.xticks(rotation=45)

plt.title(
    "Task 2 - Integrated Gradients Token Importance"
)

plt.show()


print("\n===== FINAL PERFORMANCE VISUALIZATIONS =====\n")

# ================================================================
# 23. FINAL PERFORMANCE VISUALIZATIONS
# ================================================================

import matplotlib.pyplot as plt
import seaborn as sns


# ================================================================
# 23.1 BERTScore Plot
# ================================================================

plt.figure(figsize=(10, 5))

sns.barplot(
    data=bert_summary,
    x='prompt_type',
    y='bert_f1',
    palette=prompt_palette
)

plt.title("TASK2:  Customer Communication Drafting - BERTScore Comparison")

plt.xticks(rotation=45)

plt.show()


# ================================================================
# 23.2 POLITENESS Plot
# ================================================================

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='politeness',
    palette=prompt_palette
)


plt.title(
    "Task 2 - Politeness Comparison"
)

plt.xticks(rotation=45)

plt.show()


# ================================================================
# 23.3 PROFESSIONALISM Plot
# ================================================================

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='professionalism',
    palette=prompt_palette
)

plt.title(
    "Task 2 - Professionalism Comparison"
)

plt.xticks(rotation=45)

plt.show()

# ================================================================
# 23.4 COMPLIANCE RISK Plot
# ================================================================

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='compliance_risk'
)

plt.title(
    "Task 2 - Compliance Risk Comparison"
)

plt.xticks(rotation=45)

plt.show()

# ================================================================
# 23.5 CLARITY Plot
# ================================================================

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='clarity',
    palette=prompt_palette
)

plt.title(
    "Task 2 - Clarity Comparison"
)

plt.xticks(rotation=45)

plt.show()

# ================================================================
# 23.6 Hallucination Plot
# ================================================================

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='hallucination',
    palette=prompt_palette
)

plt.title(
    "Task 2 - Hallucination Comparison"
)

plt.xticks(rotation=45)

plt.show()


# ================================================================
# 24. EXPORT
# ================================================================

print("\n===== EXPORTING TABLES =====\n")

task2_df.to_excel(
    f"{MODEL_SHORT}_task2_outputs.xlsx",
    index=False
)

bert_summary.to_excel(
    f"{MODEL_SHORT}_task2_bertscore_summary.xlsx",
    index=False
)

final_task2_summary.to_excel(
    f"{MODEL_SHORT}_task2_final_summary.xlsx",
    index=False
)

lime_df.to_excel(
    f"{MODEL_SHORT}_task2_lime_importance.xlsx",
    index=False
)

shap_df.to_excel(
    f"{MODEL_SHORT}_task2_shap_importance.xlsx",
    index=False
)

ig_df.to_excel(
    f"{MODEL_SHORT}_task2_ig_importance.xlsx",
    index=False
)

print("Tables exported successfully.")

print("\n===== EXPORTING FIGURES =====\n")

FIG_DPI = 600

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='professionalism',
    palette=prompt_palette
)

plt.title("Task 2 - Professionalism Comparison")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_professionalism.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='clarity',
    palette=prompt_palette
)

plt.title("Task 2 - Clarity Comparison")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_clarity.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='compliance_risk',
    palette=prompt_palette
)

plt.title("Task 2 - Compliance Risk Comparison")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_compliance_risk.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task2_summary,
    x='prompt_type',
    y='hallucination',
    palette=prompt_palette
)

plt.title("Task 2 - Hallucination Comparison")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_hallucination.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=bert_summary,
    x='prompt_type',
    y='bert_f1',
    palette=prompt_palette
)

plt.title("Task 2 - BERTScore Comparison")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_bertscore.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    lime_plot_df_task2['token'],
    lime_plot_df_task2['importance'],
    color=lime_colors
)

plt.xticks(rotation=45)

plt.title(
    "Task 2 - LIME Token Importance"
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_lime.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    top_shap_task2['token'],
    top_shap_task2['abs_importance'],
    color=shap_colors
)

plt.xticks(rotation=45)

plt.title(
    "Task 2 - SHAP Token Importance"
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_shap.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    top_ig_task2['token'],
    top_ig_task2['abs_importance'],
    color=ig_colors
)

plt.xticks(rotation=45)

plt.title(
    "Task 2 - Integrated Gradients Token Importance"
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_integrated_gradients.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(12,6))

sns.heatmap(
    heatmap_data,
    annot=True,
    cmap='Blues'
)

plt.title(
    "Task 2 Prompt Performance Heatmap"
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task2_heatmap.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

print("\n===== EXPORT COMPLETED =====\n")

print("\n===== TASK 2 EXECUTION COMPLETED =====\n")

print(
    "Total Outputs Generated:",
    len(task2_df)
)

print(
    "Total Prompt Types:",
    task2_df['prompt_type'].nunique()
)

print(
    "Model Used:",
    MODEL_NAME
)

```
