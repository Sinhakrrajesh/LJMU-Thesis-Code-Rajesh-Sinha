```python
print("\n===== Importing Required Libraries =====\n")
# ================================================================
# 2. Importing Required Libraries
# ================================================================

# ================================================================
# 2.1 Core Python & Utility Libraries
# ================================================================
import os
import re
import time
import random
import warnings
warnings.filterwarnings('ignore')

# ================================================================
# 2.2 Data Handling & Processing Libraries
# ================================================================
import pandas as pd
import numpy as np

# ================================================================
# 2.3 Progress Monitoring Libraries
# ================================================================
from tqdm import tqdm

# ================================================================
# 2.4 Data Visualization (EDA + Results)
# ================================================================
import matplotlib.pyplot as plt
import seaborn as sns

# ================================================================
# 2.5 LLM / Model APIs
# ================================================================
from groq import Groq
from openai import OpenAI

# ================================================================
# 2.6 Evaluation Metrics
# ================================================================
from bert_score import score as bertscore
from rouge_score import rouge_scorer
from sklearn.metrics import accuracy_score

# ================================================================
# 2.7 Explainable AI
# ================================================================
from lime.lime_text import LimeTextExplainer
import shap
import lime
from captum.attr import IntegratedGradients

# ================================================================
# 2.8 LIBRARY VERSIONS
# ================================================================

print("===== LIBRARY VERSIONS =====\n")

from importlib.metadata import version, PackageNotFoundError

def get_version(pkg_name):
    try:
        return version(pkg_name)
    except PackageNotFoundError:
        return "Not installed"
    except Exception:
        return "Version not found"

# Core Python
import platform
print("Python:", platform.python_version())

# Data
print("pandas:", pd.__version__)
print("numpy:", np.__version__)

# Progress
import tqdm

print("tqdm:", tqdm.__version__)

# Visualization
import matplotlib
print("matplotlib:", matplotlib.__version__)
print("seaborn:", sns.__version__)

# LLM APIs
print("groq:", get_version("groq"))
print("openai:", get_version("openai"))

# Evaluation
print("rouge-score:", get_version("rouge-score"))
print("scikit-learn:", get_version("scikit-learn"))

# Explainability
print("lime:", get_version("lime"))
print("shap:", get_version("shap"))

# Built-in
print("re: built-in (no version)")

# ================================================================
# 3. GROQ API key Loading
# ================================================================
print("\n===== GROQ API key Loading =====\n")

from google.colab import userdata

GROQ_KEYS = [
    userdata.get('GROQ_API_KEY1'),
    userdata.get('GROQ_API_KEY2'),
    userdata.get('GROQ_API_KEY3'),
    userdata.get('GROQ_API_KEY4'),
    userdata.get('GROQ_API_KEY5'),
    userdata.get('GROQ_API_KEY6'),
    userdata.get('GROQ_API_KEY7'),
    userdata.get('GROQ_API_KEY8'),
    userdata.get('GROQ_API_KEY9'),
    userdata.get('GROQ_API_KEY10'),
    userdata.get('GROQ_API_KEY11'),
    userdata.get('GROQ_API_KEY12')
]

print("\n===== GROQ API keys Loaded Successfully from Colab Secrets =====\n")


print("\n===== LOADING COMMON BFSI DATASET =====\n")

# ================================================================
# 4. LOAD PREPROCESSED BFSI DATASET
# ================================================================

# The dataset used in this experiment was preprocessed
# using a separate dataset preparation pipeline.
# The cleaned and validated dataset is loaded directly
# for prompt engineering experiments.

# ================================================================
# LOAD COMMON DATASET
# ================================================================

df = pd.read_csv(

    "/content/common_bfsi_dataset.csv"
)

print(
    "Dataset loaded successfully"
)

print(
    "Dataset Shape:",
    df.shape
)

# ================================================================
# DISPLAY SAMPLE RECORDS
# ================================================================

print("\n===== SAMPLE RECORDS =====\n")

print(
    df[[
        'Product',
        'Issue',
        'clean_text'
    ]].head(3)
)

# ================================================================
# DATASET VALIDATION
# ================================================================

print("\n===== DATASET VALIDATION =====\n")

print(
    "Null values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Columns:",
    list(df.columns)
)

print("\n===== FINAL EXPERIMENT DATASET =====\n")

print(
    "Dataset Shape:",
    df.shape
)

print(
    "Total Records:",
    len(df)
)



# ================================================================
# 5. Model and Task
# ================================================================

print("\n===== MODEL & TASK =====\n")
print("\n=====   MODEL:  openai/gpt-oss-20b  =====\n")
print("\n=====   TASK-3:  Decision Explanation and Justification   =====\n")

MODEL_NAME = "openai/gpt-oss-20b"
MODEL_SHORT = "gptoss20b"
TASK_NAME = "Task-3: Decision Explanation and Justification"

MODELS = [
    "openai/gpt-oss-20b"
]

current_key_index = 0
blocked_keys = {}
error_counter = 0

def run_model(prompt, model_name, retries=5):

    global current_key_index
    global error_counter

    for attempt in range(retries):

        try:

            start_index = current_key_index

            while True:

                key = GROQ_KEYS[current_key_index]

                if key not in blocked_keys or time.time() > blocked_keys[key]:
                    break

                current_key_index = (
                    current_key_index + 1
                ) % len(GROQ_KEYS)

                if current_key_index == start_index:

                    time.sleep(60)

            client = Groq(api_key=key)

            current_key_index = (
                current_key_index + 1
            ) % len(GROQ_KEYS)

            time.sleep(random.uniform(1, 3))

            response = client.chat.completions.create(

                model=model_name,

                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],

                temperature=0.3,

                top_p=0.9,

                max_tokens=300,

                timeout=60
            )

            output = response.choices[0].message.content

            if output is None:
                return "ERROR"

            return output

        except Exception as e:

            error_counter += 1

            blocked_keys[key] = time.time() + 120

            time.sleep(5)

    return "ERROR"


# ================================================================
# 6. Generic Experiment Engine
# ================================================================

def run_experiment(df, prompts):

    results = []

    for i, row in tqdm.tqdm(
        df.iterrows(),
        total=len(df)
    ):

        text = row['clean_text']

        for p_type, func in prompts.items():

            prompt = func(text)

            output = run_model(
                prompt,
                MODEL_NAME
            )

            results.append({

                "id": i,

                "model": MODEL_NAME,

                "prompt_type": p_type,

                "input_text": text,

                "output": output
            })

        # ========================================================
        # CHECKPOINT SAVE EVERY 25 RECORDS
        # ========================================================

        if i % 25 == 0 and i != 0:

            checkpoint_df = pd.DataFrame(results)

            checkpoint_df.to_excel(

                f"{TASK_NAME}_{MODEL_SHORT}_checkpoint.xlsx",

                index=False
            )

    return pd.DataFrame(results)

# ================================================================
# 7. PROMPT STRATEGIES
# ================================================================


# ------------------------------------------------
# 7.1 ZERO SHOT
# ------------------------------------------------

def zero_shot_decision_prompt(text):

    return f"""

Analyze the banking complaint and provide a decision
with clear justification.

Complaint:
{text}

Output:
- Decision
- Justification

"""

# ------------------------------------------------
# 7.2 FEW SHOT
# ------------------------------------------------

def few_shot_decision_prompt(text):

    example = """

Complaint:
Customer was charged duplicate fees.

Decision:
Refund recommended.

Justification:
Duplicate charges may indicate billing errors
affecting customer fairness and compliance standards.

"""

    return f"""

Example:
{example}

Now analyze the following complaint.

Complaint:
{text}

Output:
- Decision
- Justification

"""

# ------------------------------------------------
# 7.3 ROLE BASED
# ------------------------------------------------

def role_decision_prompt(text):

    return f"""

You are a senior BFSI compliance officer.

Review the complaint carefully and provide:

1. Decision
2. Compliance justification
3. Risk assessment
4. Recommended action

Complaint:
{text}

Output:
- Decision
- Compliance Justification
- Risk Assessment
- Recommended Action

"""

# ------------------------------------------------
# 7.4 INSTRUCTION PROMPT
# ------------------------------------------------

def instruction_decision_prompt(text):

    return f"""

Follow these steps carefully:

1. Identify the main issue
2. Determine financial risk severity
3. Analyze compliance implications
4. Make a justified decision
5. Recommend next action

Complaint:
{text}

Output:
- Decision
- Explanation
- Recommended Action

"""

# ------------------------------------------------
# 7.5 CONSTRAINT PROMPT
# ------------------------------------------------

def constraint_decision_prompt(text):

    return f"""

Rules:

- Use ONLY facts from the complaint
- Do NOT assume missing information
- Keep justification evidence-based
- Avoid unsupported legal conclusions
- Avoid speculative recommendations

Complaint:
{text}

Output:
- Decision
- Justification

"""

# ------------------------------------------------
# 7.6 CHAIN OF THOUGHT (CoT)
# ------------------------------------------------

def cot_decision_prompt(text):

    return f"""

You are a BFSI regulatory decision expert.

Step 1:
Identify the primary complaint issue.

Step 2:
Determine customer impact.

Step 3:
Evaluate financial and compliance risks.

Step 4:
Decide appropriate resolution.

Step 5:
Provide detailed justification.

Complaint:
{text}

Output:
- Decision
- Detailed Justification
- Compliance Consideration
- Recommended Action

"""


task3_prompts = {

    "zero_shot": zero_shot_decision_prompt,

    "few_shot": few_shot_decision_prompt,

    "role_based": role_decision_prompt,

    "instruction": instruction_decision_prompt,

    "constraint": constraint_decision_prompt,

    "cot": cot_decision_prompt
}

print("\n===== TASK-3:  Decision Explanation and Justification - EXPERIMENT EXECUTION =====\n")

# ================================================================
# 8. EXECUTION
# ================================================================

task3_df = run_experiment(df, task3_prompts)

# ================================================================
# TASK 3.5 METRICS
# ================================================================

# Logical Reasoning
def logic_score(text):

    text = str(text).lower()

    logic_terms = [

        "decision",
        "justification",
        "rationale",
        "reason",
        "analysis",
        "assessment",
        "evaluation",
        "finding",
        "findings",
        "determine",
        "determine financial",
        "step",
        "issue",
        "recommend",
        "recommendation",
        "because",
        "therefore",
        "based on",
        "conclusion"
    ]

    matches = sum(
        term in text
        for term in logic_terms
    )

    return round(
        matches / len(logic_terms),
        4
    )


# Consistency
def consistency_score(text):

    text = str(text).lower()

    consistency_terms = [

        "consumer",
        "complaint",
        "credit",
        "identity",
        "fraud",
        "documentation",
        "verification",
        "investigation",
        "report",
        "account"
    ]

    matches = sum(
        term in text
        for term in consistency_terms
    )

    return round(
        matches / len(consistency_terms),
        4
    )


# Traceability
def traceability(text):

    keywords = [
        "complaint",
        "customer",
        "account",
        "issue",
        "records"
    ]

    return sum(
        k in text.lower()
        for k in keywords
    ) / len(keywords)


# Regulatory Alignment
def regulatory_alignment(text):

    text = str(text).lower()

    regulatory_terms = [

        "fcra",
        "fair credit reporting act",
        "15 u.s.c",
        "605(b)",
        "1681c",
        "statutory",
        "compliance",
        "regulation",
        "legal",
        "consumer protection",
        "identity theft",
        "fraud alert",
        "credit reporting agency",
        "business days",
        "documentation"
    ]

    matches = sum(
        term in text
        for term in regulatory_terms
    )

    return round(
        matches / len(regulatory_terms),
        4
    )


# Hallucination
def decision_hallucination(text):

    text = str(text).lower()

    high_risk_terms = [

        "guaranteed",
        "definitely",
        "certainly",
        "100%",
        "confirmed",
        "approved",
        "fully secure",
        "risk-free",
        "without failure",
        "permanently resolved"
    ]

    medium_risk_terms = [

        "verified",
        "authorized",
        "completely safe",
        "always successful",
        "no compliance risk"
    ]

    low_risk_terms = [

        "assume",
        "possibly",
        "might",
        "likely"
    ]

    high = sum(
        term in text
        for term in high_risk_terms
    )

    medium = sum(
        term in text
        for term in medium_risk_terms
    )

    low = sum(
        term in text
        for term in low_risk_terms
    )

    score = (

        high * 0.6 +

        medium * 0.3 +

        low * 0.1

    ) / 5

    return min(score, 1)


# ================================================================
# REMOVE EMPTY / ERROR OUTPUTS
# ================================================================

task3_df = task3_df[
    task3_df['output'].notna()
]

task3_df = task3_df[
    task3_df['output'].str.strip() != ""
]

task3_df = task3_df[
    task3_df['output'] != "ERROR"
]

task3_df = task3_df.reset_index(drop=True)

print("Remaining rows:", len(task3_df))

#  APPLY TASK 3 METRICS
task3_df['logic'] = task3_df[
    'output'
].apply(logic_score)

task3_df['consistency'] = task3_df[
    'output'
].apply(consistency_score)

task3_df['traceability'] = task3_df[
    'output'
].apply(traceability)

task3_df['regulatory_alignment'] = task3_df[
    'output'
].apply(regulatory_alignment)

task3_df['hallucination'] = task3_df[
    'output'
].apply(decision_hallucination)


# ================================================================
# TASK 3.6 SUMMARY
# ================================================================

summary_task3 = task3_df.groupby(
    ['model', 'prompt_type']
).agg({

    'logic': 'mean',
    'consistency': 'mean',
    'traceability': 'mean',
    'regulatory_alignment': 'mean',
    'hallucination': 'mean'

}).reset_index()

display(summary_task3)


# ================================================================
# TASK 3.7 BERTScore
# ================================================================

from bert_score import score as bertscore

print("Running Task 3 BERTScore...")

P, R, F1 = bertscore(
    task3_df['output'].astype(str).tolist(),
    task3_df['input_text'].astype(str).tolist(),
    lang='en',
    model_type='roberta-large',
    verbose=True
)

task3_df['bert_precision'] = P.numpy()
task3_df['bert_recall'] = R.numpy()
task3_df['bert_f1'] = F1.numpy()

# Summary
bert_summary_task3 = task3_df.groupby(
    ['model', 'prompt_type']
).agg({
    'bert_f1': 'mean'
}).reset_index()

display(bert_summary_task3)

# ================================================================
# TASK 3.8 SAMPLE OUTPUT ANALYSIS
# ================================================================


for p in task3_df['prompt_type'].unique():

    print(f"\n\n===== TASK 3 : {p.upper()} =====\n")

    sample_output = task3_df[
        task3_df['prompt_type'] == p
    ]['output'].iloc[0]

    print(sample_output)

# ================================================================
# TASK 3.9 LIME
# ================================================================

from lime.lime_text import LimeTextExplainer

print("Running Task 3 LIME...")

explainer = LimeTextExplainer(
    class_names=['decision']
)

# Decision reasoning predictor
def predictor(texts):

    probs = []

    for text in texts:

        score = (

            logic_score(text) +

            consistency_score(text) +

            traceability(text) +

            regulatory_alignment(text)

        ) / 4

        score = min(max(score, 0), 1)

        probs.append([

            1 - score,

            score

        ])

    return np.array(probs)

sample_text = task3_df['output'].iloc[0]

exp = explainer.explain_instance(
    sample_text,
    predictor,
    num_features=10
)

exp.show_in_notebook(text=True)


# ================================================================
# TASK 3.10 LIME FEATURE IMPORTANCE
# ================================================================

lime_features_task3 = exp.as_list()

lime_df_task3 = pd.DataFrame(
    lime_features_task3,
    columns=['word', 'importance']
)

display(lime_df_task3)

# ================================================================
# TASK 3.11 SHAP
# ================================================================

print("Running Task 3 SHAP...")

def shap_predict(texts):

    scores = []

    for text in texts:

        score = (

            logic_score(text) +

            consistency_score(text) +

            traceability(text) +

            regulatory_alignment(text)

        ) / 4

        scores.append(score)

    return np.array(scores)

explainer = shap.Explainer(

    shap_predict,

    shap.maskers.Text()
)

sample_text = task3_df['output'].iloc[0]

shap_values = explainer([sample_text])

shap.plots.text(shap_values[0])

# ================================================================
# SHAP FEATURE IMPORTANCE
# ================================================================

shap_tokens_task3 = np.array(
    shap_values.data[0]
)

shap_scores_task3 = np.array(
    shap_values.values[0]
)

shap_df_task3 = pd.DataFrame({
    'token': shap_tokens_task3,
    'importance': shap_scores_task3
})

shap_df_task3 = shap_df_task3.dropna()

shap_df_task3['token'] = (

    shap_df_task3['token']

    .astype(str)

    .str.replace(
        "##",
        "",
        regex=False
    )
)

shap_df_task3 = shap_df_task3[

    shap_df_task3['token'].str.len() > 2
]

shap_df_task3['abs_importance'] = (
    shap_df_task3['importance'].abs()
)

shap_df_task3 = shap_df_task3.sort_values(
    by='abs_importance',
    ascending=False
)

display(shap_df_task3.head(20))

# ================================================================
# TASK 3.12 INTEGRATED GRADIENTS (IG)
# ================================================================

print("Running Task 3 Integrated Gradients...")

import torch

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

from captum.attr import IntegratedGradients

# Load local surrogate transformer model
tokenizer = AutoTokenizer.from_pretrained(
    "distilbert-base-uncased-finetuned-sst-2-english"
)

model = AutoModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased-finetuned-sst-2-english"
)

model.eval()

# Sample reasoning output
sample_text = task3_df['output'].iloc[0]

# Tokenize
inputs = tokenizer(
    sample_text,
    return_tensors="pt",
    truncation=True,
    padding=True
)

# Get embedding layer
embedding_layer = model.distilbert.embeddings.word_embeddings

# Forward function
def forward_func_emb(inputs_embeds, attention_mask):

    outputs = model(
        inputs_embeds=inputs_embeds,
        attention_mask=attention_mask
    )

    return outputs.logits[:, 1]

# Initialize IG
ig = IntegratedGradients(forward_func_emb)

# Input embeddings
input_embeddings = embedding_layer(
    inputs['input_ids']
)

# Compute attributions
attributions, delta = ig.attribute(
    inputs=input_embeddings,
    additional_forward_args=(
        inputs['attention_mask'],
    ),
    return_convergence_delta=True
)

# Tokens
tokens = tokenizer.convert_ids_to_tokens(
    inputs['input_ids'][0]
)

# Scores
scores = attributions.sum(dim=-1).squeeze(0).detach().numpy()

# Create dataframe
ig_df_task3 = pd.DataFrame({
    'token': tokens,
    'importance': scores
})

ig_df_task3['token'] = (

    ig_df_task3['token']

    .str.replace(
        "##",
        "",
        regex=False
    )
)

# Remove special tokens
ig_df_task3 = ig_df_task3[
    ~ig_df_task3['token'].isin(['[CLS]', '[SEP]'])
]

# Remove very small tokens
ig_df_task3 = ig_df_task3[

    ig_df_task3['token'].str.len() > 2
]


# Absolute importance
ig_df_task3['abs_importance'] = (
    ig_df_task3['importance'].abs()
)

# Sort
ig_df_task3 = ig_df_task3.sort_values(
    by='abs_importance',
    ascending=False
)

display(ig_df_task3.head(20))


# ================================================================
# FINAL TASK 3.14 SUMMARY
# ================================================================

final_task3_summary = task3_df.groupby(
    ['model', 'prompt_type']
).agg({

    'logic': 'mean',
    'consistency': 'mean',
    'traceability': 'mean',
    'regulatory_alignment': 'mean',
    'hallucination': 'mean',
    'bert_f1': 'mean'

}).reset_index()

display(final_task3_summary)

# ================================================================
# TASK 3 PERFORMANCE HEATMAP
# ================================================================

print("\n===== TASK 3 PERFORMANCE HEATMAP =====\n")

heatmap_data_task3 = final_task3_summary.set_index(

    'prompt_type'

)[[

    'logic',

    'consistency',

    'traceability',

    'regulatory_alignment',

    'hallucination',

    'bert_f1'
]]

plt.figure(figsize=(12,6))

sns.heatmap(

    heatmap_data_task3,

    annot=True,

    cmap='YlGnBu'

)

plt.title(

    "Task 3 - Prompt Performance Heatmap"
)

plt.tight_layout()

plt.show()
# ================================================================
# TASK 3.15 VISUALIZATION
# ================================================================

plt.figure(figsize=(12,6))

# ================================================================
# COLOR PALETTE
# ================================================================

prompt_palette = {

    'zero_shot': '#1f77b4',
    'few_shot': '#ff7f0e',
    'role_based': '#2ca02c',
    'instruction': '#d62728',
    'constraint': '#9467bd',
    'cot': '#8c564b'
}

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='logic',
    palette=prompt_palette,
    hue='prompt_type',
    legend=False
)

plt.title("Task 3 - Logic Comparison")

plt.xticks(rotation=45)

plt.show()


# ================================================================
# VISUALIZATION - CONSISTENCY
# ================================================================

plt.figure(figsize=(12,6))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='consistency',
    palette=prompt_palette,
    hue='prompt_type',
    legend=False
)

plt.title("Task 3 - Consistency Comparison")

plt.xticks(rotation=45)

plt.show()

# ================================================================
# VISUALIZATION - TRACEABILITY
# ================================================================

plt.figure(figsize=(12,6))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='traceability',
    palette=prompt_palette,
    hue='prompt_type',
    legend=False
)

plt.title("Task 3 - Traceability Comparison")

plt.xticks(rotation=45)

plt.show()

# ================================================================
# VISUALIZATION - REGULATORY ALIGNMENT
# ================================================================

plt.figure(figsize=(12,6))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='regulatory_alignment',
    palette=prompt_palette,
    hue='prompt_type',
    legend=False
)


plt.title("Task 3 - Regulatory Alignment Comparison")

plt.xticks(rotation=45)

plt.show()

# ================================================================
# VISUALIZATION - HALLUCINATION
# ================================================================

plt.figure(figsize=(12,6))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='hallucination',
    palette=prompt_palette,
    hue='prompt_type',
    legend=False
)


plt.title("Task 3 - Hallucination Comparison")

plt.xticks(rotation=45)

plt.show()

# ================================================================
# VISUALIZATION - BERTScore
# ================================================================

plt.figure(figsize=(12,6))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='bert_f1',
    palette=prompt_palette,
    hue='prompt_type',
    legend=False
)

plt.title("Task 3 - BERTScore Comparison")

plt.xticks(rotation=45)

plt.show()

# ================================================================
# VISUALIZATION - LIME
# ================================================================

plt.figure(figsize=(10,5))

lime_plot_df_task3 = lime_df_task3.sort_values(
    by='importance',
    ascending=False
)

lime_colors = plt.cm.tab20(

    np.linspace(
        0,
        1,
        len(lime_plot_df_task3)
    )
)

plt.bar(
    lime_plot_df_task3['word'],
    lime_plot_df_task3['importance'],
    color=lime_colors
)

plt.xticks(rotation=45)

plt.title("Task 3 - LIME Token Importance")

plt.show()

# ================================================================
# VISUALIZATION - SHAP
# ================================================================

top_shap_task3 = shap_df_task3.head(10)

plt.figure(figsize=(10,5))

shap_colors = plt.cm.Set3(

    np.linspace(
        0,
        1,
        len(top_shap_task3)
    )
)

plt.bar(
    top_shap_task3['token'],
    top_shap_task3['abs_importance'],
    color=shap_colors
)

plt.xticks(rotation=45)

plt.title("Task 3 - SHAP Token Attribution")

plt.show()

# ================================================================
# VISUALIZATION - IG
# ================================================================

top_ig_task3 = ig_df_task3.head(10)

plt.figure(figsize=(10,5))

ig_colors = [

    '#1f77b4',
    '#ff7f0e',
    '#2ca02c',
    '#d62728',
    '#9467bd',
    '#8c564b',
    '#e377c2',
    '#7f7f7f',
    '#bcbd22',
    '#17becf'
]

plt.bar(
    top_ig_task3['token'],
    top_ig_task3['abs_importance'],
    color=ig_colors[:len(top_ig_task3)]
)

plt.xticks(rotation=45)

plt.title("Task 3 - Integrated Gradients Token Importance")

plt.show()


# ================================================================
# EXPORT TABLES
# ================================================================

print("\n===== EXPORTING TABLES =====\n")

task3_df.to_excel(
    f"{MODEL_SHORT}_task3_outputs.xlsx",
    index=False
)

summary_task3.to_excel(
    f"{MODEL_SHORT}_task3_summary.xlsx",
    index=False
)

bert_summary_task3.to_excel(
    f"{MODEL_SHORT}_task3_bertscore_summary.xlsx",
    index=False
)

lime_df_task3.to_excel(
    f"{MODEL_SHORT}_task3_lime_importance.xlsx",
    index=False
)

shap_df_task3.to_excel(
    f"{MODEL_SHORT}_task3_shap_importance.xlsx",
    index=False
)

ig_df_task3.to_excel(
    f"{MODEL_SHORT}_task3_ig_importance.xlsx",
    index=False
)

heatmap_data_task3.to_excel(
    f"{MODEL_SHORT}_task3_heatmap_data.xlsx"
)

print("Tables exported successfully.")

# ================================================================
# EXPORT FIGURES
# ================================================================

FIG_DPI = 600

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='logic',
    palette=prompt_palette
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_logic.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='consistency',
    palette=prompt_palette
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_consistency.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='traceability',
    palette=prompt_palette
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_traceability.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='regulatory_alignment',
    palette=prompt_palette
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_regulatory_alignment.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=final_task3_summary,
    x='prompt_type',
    y='hallucination',
    palette=prompt_palette
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_hallucination.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=bert_summary_task3,
    x='prompt_type',
    y='bert_f1',
    palette=prompt_palette
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_bertscore.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    lime_plot_df_task3['word'],
    lime_plot_df_task3['importance'],
    color=lime_colors
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_lime.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    top_shap_task3['token'],
    top_shap_task3['abs_importance'],
    color=shap_colors
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_shap.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    top_ig_task3['token'],
    top_ig_task3['abs_importance'],
    color=ig_colors[:len(top_ig_task3)]
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_integrated_gradients.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(12,6))

sns.heatmap(
    heatmap_data_task3,
    annot=True,
    cmap='YlGnBu'
)

plt.title(
    "Task 3 - Prompt Performance Heatmap"
)

plt.tight_layout()

plt.savefig(
    f"{MODEL_SHORT}_task3_heatmap.png",
    dpi=FIG_DPI,
    bbox_inches='tight'
)

plt.close()


print("\n===== TASK 3 EXECUTION COMPLETED =====\n")

print(
    "Total Outputs Generated:",
    len(task3_df)
)

print(
    "Total Prompt Types:",
    task3_df['prompt_type'].nunique()
)

print(
    "Model Used:",
    MODEL_NAME
)

print(
    "Total API Errors:",
    error_counter
)
```
