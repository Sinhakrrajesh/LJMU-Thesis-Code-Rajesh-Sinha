```python
print("\n===== Importing Required Libraries =====\n")
# ================================================================
# 2. Importing Required Libraries
# ================================================================

# ================================================================
# 2.1 Core Python & Utility Libraries
# ================================================================
import os
import re
import time
import random
import warnings
warnings.filterwarnings('ignore')

# ================================================================
# 2.2 Data Handling & Processing Libraries
# ================================================================
import pandas as pd
import numpy as np

# ================================================================
# 2.3 Progress Monitoring Libraries
# ================================================================
from tqdm import tqdm

# ================================================================
# 2.4 Data Visualization (EDA + Results)
# ================================================================
import matplotlib.pyplot as plt
import seaborn as sns

# ================================================================
# 2.5 LLM / Model APIs
# ================================================================
from groq import Groq
from openai import OpenAI

# ================================================================
# 2.6 Evaluation Metrics
# ================================================================
from bert_score import score as bertscore
from rouge_score import rouge_scorer
from sklearn.metrics import accuracy_score

# ================================================================
# 2.7 Explainable AI
# ================================================================
from lime.lime_text import LimeTextExplainer
import shap
import lime
from captum.attr import IntegratedGradients

# ================================================================
# 2.8 LIBRARY VERSIONS
# ================================================================

print("===== LIBRARY VERSIONS =====\n")

from importlib.metadata import version, PackageNotFoundError

def get_version(pkg_name):
    try:
        return version(pkg_name)
    except PackageNotFoundError:
        return "Not installed"
    except Exception:
        return "Version not found"

# Core Python
import platform
print("Python:", platform.python_version())

# Data
print("pandas:", pd.__version__)
print("numpy:", np.__version__)

# Progress
import tqdm

print("tqdm:", tqdm.__version__)

# Visualization
import matplotlib
print("matplotlib:", matplotlib.__version__)
print("seaborn:", sns.__version__)

# LLM APIs
print("groq:", get_version("groq"))
print("openai:", get_version("openai"))

# Evaluation
print("rouge-score:", get_version("rouge-score"))
print("scikit-learn:", get_version("scikit-learn"))

# Explainability
print("lime:", get_version("lime"))
print("shap:", get_version("shap"))

# Built-in
print("re: built-in (no version)")

# ================================================================
# 3. GROQ API key Loading
# ================================================================
print("\n===== GROQ API key Loading =====\n")

from google.colab import userdata

GROQ_KEYS = [
    userdata.get('GROQ_API_KEY1'),
    userdata.get('GROQ_API_KEY2'),
    userdata.get('GROQ_API_KEY3'),
    userdata.get('GROQ_API_KEY4'),
    userdata.get('GROQ_API_KEY5'),
    userdata.get('GROQ_API_KEY6'),
    userdata.get('GROQ_API_KEY7'),
    userdata.get('GROQ_API_KEY8'),
    userdata.get('GROQ_API_KEY9'),
    userdata.get('GROQ_API_KEY10'),
    userdata.get('GROQ_API_KEY11'),
    userdata.get('GROQ_API_KEY12')
]

print("\n===== GROQ API keys Loaded Successfully from Colab Secrets =====\n")


print("\n===== LOADING COMMON BFSI DATASET =====\n")

# ================================================================
# 4. LOAD PREPROCESSED BFSI DATASET
# ================================================================

# The dataset used in this experiment was preprocessed
# using a separate dataset preparation pipeline.
# The cleaned and validated dataset is loaded directly
# for prompt engineering experiments.

# ================================================================
# LOAD COMMON DATASET
# ================================================================

df = pd.read_csv(

    "/content/common_bfsi_dataset.csv"
)

print(
    "Dataset loaded successfully"
)

print(
    "Dataset Shape:",
    df.shape
)

# ================================================================
# DISPLAY SAMPLE RECORDS
# ================================================================

print("\n===== SAMPLE RECORDS =====\n")

print(
    df[[
        'Product',
        'Issue',
        'clean_text'
    ]].head(3)
)

# ================================================================
# DATASET VALIDATION
# ================================================================

print("\n===== DATASET VALIDATION =====\n")

print(
    "Null values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Columns:",
    list(df.columns)
)

print("\n===== FINAL EXPERIMENT DATASET =====\n")

print(
    "Dataset Shape:",
    df.shape
)

print(
    "Total Records:",
    len(df)
)



# ================================================================
# 5. Model and Task
# ================================================================

print("\n===== MODEL & TASK =====\n")
print("\n=====   MODEL:  openai/gpt-oss-120b   =====\n")
print("\n=====   TASK1:  Risk Analysis & Complaint Summarization   =====\n")

MODEL_NAME = "openai/gpt-oss-120b"
MODEL_SHORT = "gptoss120b"

TASK_NAME = "Task-1: Risk Analysis & Complaint Summarization"

MODELS = [
    "openai/gpt-oss-120b"
]

current_key_index = 0
blocked_keys = {}
error_counter = 0

def run_model(prompt, model_name, retries=5):

    global current_key_index
    global error_counter

    for attempt in range(retries):

        try:

            start_index = current_key_index

            while True:

                key = GROQ_KEYS[current_key_index]

                if key not in blocked_keys or time.time() > blocked_keys[key]:
                    break

                current_key_index = (
                    current_key_index + 1
                ) % len(GROQ_KEYS)

                if current_key_index == start_index:

                    time.sleep(60)

            client = Groq(api_key=key)

            current_key_index = (
                current_key_index + 1
            ) % len(GROQ_KEYS)

            time.sleep(random.uniform(1, 3))

            response = client.chat.completions.create(

                model=model_name,

                messages=[
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],

                temperature=0.3,

                top_p=0.9,

                max_tokens=300,

                timeout=60
            )

            output = response.choices[0].message.content

            if output is None:
                return "ERROR"

            return output

        except Exception as e:

            error_counter += 1

            blocked_keys[key] = time.time() + 120

            time.sleep(5)

    return "ERROR"


# ================================================================
# 6. Generic Experiment Engine
# ================================================================

def run_experiment(df, prompts):

    results = []

    for i, row in tqdm.tqdm(
        df.iterrows(),
        total=len(df)
    ):

        text = row['clean_text']

        for p_type, func in prompts.items():

            prompt = func(text)

            output = run_model(
                prompt,
                MODEL_NAME
            )

            results.append({

                "id": i,

                "model": MODEL_NAME,

                "prompt_type": p_type,

                "input_text": text,

                "output": output
            })

        # ========================================================
        # CHECKPOINT SAVE EVERY 25 RECORDS
        # ========================================================

        if i % 25 == 0 and i != 0:

            checkpoint_df = pd.DataFrame(results)

            checkpoint_df.to_excel(

                f"{TASK_NAME}_{MODEL_SHORT}_checkpoint.xlsx",

                index=False
            )

    return pd.DataFrame(results)

# ================================================================
# 7. PROMPTS
# ================================================================


# ------------------------------------------------
# 7.1 ZERO SHOT
# ------------------------------------------------

def zero_shot_prompt(text):

    return f"""

Summarize financial risk from complaint.

Complaint:
{text}

Output:
- Risk
- Impact
- Severity

"""


# ------------------------------------------------
# 7.2 FEW SHOT
# ------------------------------------------------

def few_shot_prompt(text):

    return f"""

Example:

Complaint:
The customer was charged multiple unauthorized fees and did not receive clarification from the bank.

Risk:
Unauthorized financial charges

Impact:
Customer financial loss and dissatisfaction

Severity:
High


Now analyze the following complaint.

Complaint:
{text}

Output:
- Risk
- Impact
- Severity

"""


# ------------------------------------------------
# 7.3 ROLE BASED
# ------------------------------------------------

def role_based_prompt(text):

    return f"""

You are a senior BFSI compliance and financial risk analyst.

Analyze the customer complaint from a banking risk perspective.

Complaint:
{text}

Provide:

1. Financial Risk
2. Regulatory Risk
3. Customer Impact
4. Severity Level
5. Recommended Action

"""


# ------------------------------------------------
# 7.4 INSTRUCTION PROMPT
# ------------------------------------------------

def instruction_prompt(text):

    return f"""

Carefully analyze the following banking complaint.

Instructions:

1. Identify financial risks.
2. Identify possible compliance concerns.
3. Explain customer impact.
4. Explain operational impact.
5. Provide severity classification.

Complaint:
{text}

Return a structured analysis.

"""


# ------------------------------------------------
# 7.5 CONSTRAINT PROMPT
# ------------------------------------------------

def constraint_prompt(text):

    return f"""

Analyze the complaint strictly using ONLY information present in the complaint.

Do NOT assume facts.
Do NOT create additional events.
Do NOT speculate.

Complaint:
{text}

Output:
- Identified Risk
- Customer Impact
- Severity

"""


# ------------------------------------------------
# 7.6 CHAIN OF THOUGHT (CoT)
# ------------------------------------------------

def cot_prompt(text):

    return f"""

You are a BFSI risk analysis expert.

Analyze the complaint step-by-step.

Step 1:
Identify the primary issue in the complaint.

Step 2:
Identify financial or regulatory risks.

Step 3:
Explain customer and operational impact.

Step 4:
Determine severity level.

Step 5:
Provide final risk summary.

Complaint:
{text}

"""

task1_prompts = {
     "zero_shot": zero_shot_prompt,
    "few_shot": few_shot_prompt,
    "role_based": role_based_prompt,
    "instruction": instruction_prompt,
    "constraint": constraint_prompt,
     "cot": cot_prompt
}

print("\n===== TASK1:  Risk Analysis & Complaint Summarization - EXPERIMENT EXECUTION =====\n")

# ================================================================
# 8. EXECUTION
# ================================================================

task1_df = run_experiment(df, task1_prompts)

# ================================================================
# 9. METRICS
# ================================================================

# ================================================================
# 9.1 ROUGE
# ================================================================

from rouge_score import rouge_scorer

scorer = rouge_scorer.RougeScorer(
    ['rouge1', 'rougeL'],
    use_stemmer=True
)

def rouge_l(output, reference):

    return scorer.score(
        reference,
        output
    )['rougeL'].fmeasure


# ================================================================
# 9.2 FACTUAL SCORE
# ================================================================

def factual_score(text):

    keywords = [

        "fraud",

        "loan",

        "interest",

        "charge",

        "payment",

        "debt",

        "collector",

        "credit"
    ]

    text = text.lower()

    return sum(

        keyword in text

        for keyword in keywords

    ) / len(keywords)


# ================================================================
# 9.3 HALLUCINATION SCORE
# ================================================================

def hallucination(text):

    hallucination_terms = [

        "maybe",
        "possibly",
        "likely",
        "assume",
        "might",
        "could be",
        "appears",
        "suggests",
        "potentially",
        "estimated",
        "probable",
        "uncertain"
    ]

    text = text.lower()

    return sum(

        keyword in text

        for keyword in hallucination_terms

    ) / len(hallucination_terms)

# ================================================================
# 9.3 REGULATORY SCORE
# ================================================================

def regulatory_score(text):

    regulatory_terms = [

        "regulation",

        "compliance",

        "policy",

        "legal",

        "cfpb",

        "fdcpa",

        "tcpa",

        "consumer protection"
    ]

    text = text.lower()

    return sum(

        keyword in text

        for keyword in regulatory_terms

    ) / len(regulatory_terms)

# ================================================================
# 9.4 COMPLETENESS SCORE
# ================================================================

def completeness_score(text):

    completeness_terms = [

        "risk",

        "impact",

        "severity",

        "recommendation",

        "customer",

        "financial",

        "regulatory"
    ]

    text = text.lower()

    return sum(

        keyword in text

        for keyword in completeness_terms

    ) / len(completeness_terms)


print("\n===== OUTPUT CLEANING =====\n")

# ================================================================
# REMOVE EMPTY / ERROR OUTPUTS
# ================================================================

task1_df = task1_df[
    task1_df['output'].notna()
]

task1_df = task1_df[
    task1_df['output'].str.strip() != ""
]

task1_df = task1_df[
    task1_df['output'] != "ERROR"
]

task1_df = task1_df.reset_index(drop=True)

print("Remaining rows:", len(task1_df))


print("\n===== METRIC CALCULATION =====\n")

# ================================================================
# 10. APPLY METRICS
# ================================================================

task1_df['factual_score'] = task1_df[
    'output'
].apply(factual_score)

task1_df['hallucination'] = task1_df[
    'output'
].apply(hallucination)

task1_df['regulatory_score'] = task1_df[
    'output'
].apply(regulatory_score)

task1_df['completeness_score'] = task1_df[
    'output'
].apply(completeness_score)


# ================================================================
# 11. ROUGE-L
# ================================================================

# ================================================================
# 11.1 CREATE SIMPLE REFERENCE SUMMARY
# ================================================================

def create_reference_summary(text):

    text = str(text)

    return text[:300]


task1_df['reference_summary'] = task1_df[
    'input_text'
].apply(create_reference_summary)


# ================================================================
# 11.2 ROUGE-L USING REFERENCE SUMMARY
# ================================================================

task1_df['rouge_l'] = task1_df.apply(

    lambda x: rouge_l(

        x['output'],

        x['reference_summary']

    ),

    axis=1
)


# ================================================================
# 12. TASK SUMMARY
# ================================================================

summary_task1 = task1_df.groupby(

    ['prompt_type']

).agg({

    'factual_score': 'mean',

    'hallucination': 'mean',

    'regulatory_score': 'mean',

    'completeness_score': 'mean',

    'rouge_l': 'mean'

}).reset_index()

print(summary_task1)

# ================================================================
# 13. VISUALIZATION
# ================================================================

# ================================================================
# COLOR PALETTES
# ================================================================

prompt_palette = {

    'zero_shot': '#1f77b4',
    'few_shot': '#ff7f0e',
    'role_based': '#2ca02c',
    'instruction': '#d62728',
    'constraint': '#9467bd',
    'cot': '#8c564b'
}

plt.figure(figsize=(10, 5))

sns.barplot(

    data=summary_task1,

    x='prompt_type',

    y='factual_score',

    palette=prompt_palette
)

plt.title("TASK1:  Risk Analysis & Complaint Summarization -  Factual Score")

plt.xticks(rotation=45)

plt.show()


# ================================================================
# 14. BERTScore Evaluation
# ================================================================

print("\n===== BERTScore EVALUATION =====\n")


P, R, F1 = bertscore(

    task1_df['output'].astype(str).tolist(),

    task1_df['reference_summary'].astype(str).tolist(),

    lang='en',

    model_type='bert-base-uncased',

    verbose=True
)


# ================================================================
# 14.1 STORE SCORES
# ================================================================

task1_df['bert_precision'] = P.numpy()

task1_df['bert_recall'] = R.numpy()

task1_df['bert_f1'] = F1.numpy()


# ================================================================
# 14.2 SUMMARY
# ================================================================

bert_summary = task1_df.groupby(

    ['prompt_type']

).agg({

    'bert_f1': 'mean'

}).reset_index()


print("\n===== BERTScore Summary =====\n")

display(bert_summary)

# ================================================================
# 15. FINAL TASK 1 SUMMARY
# ================================================================

final_task1_summary = task1_df.groupby(

    ['model', 'prompt_type']

).agg({

    'factual_score': 'mean',

    'hallucination': 'mean',

    'regulatory_score': 'mean',

    'completeness_score': 'mean',

    'rouge_l': 'mean',

    'bert_f1': 'mean'

}).reset_index()


print("\n===== FINAL TASK1:  Risk Analysis & Complaint Summarization -  SUMMARY =====\n")

display(final_task1_summary)


print("\n===== PERFORMANCE HEATMAP =====\n")

heatmap_data = final_task1_summary.set_index(

    'prompt_type'

)[[

    'factual_score',

    'hallucination',

    'regulatory_score',

    'completeness_score',

    'rouge_l',

    'bert_f1'
]]

plt.figure(figsize=(12,6))

sns.heatmap(

    heatmap_data,

    annot=True,

    cmap='Blues'
)

plt.title("TASK1:  Risk Analysis & Complaint Summarization - Prompting Performance Heatmap")

plt.show()

print("\n===== SAMPLE OUTPUT ANALYSIS =====\n")

# ================================================================
# 16. SAMPLE OUTPUT ANALYSIS
# ================================================================

for p in task1_df['prompt_type'].unique():

    print(f"\n\n===== TASK1 : {p.upper()} =====\n")

    prompt_rows = task1_df[
        task1_df['prompt_type'] == p
    ]

    if len(prompt_rows) > 0:

        sample_output = prompt_rows[
            'output'
        ].iloc[0]

        print(sample_output[:1500])

    else:

        print("No output available")

    print("\n" + "="*80)

print("\n===== LIME EXPLAINABILITY =====\n")

# ================================================================
# 17. LIME EXPLAINABILITY
# ================================================================

print("Running LIME Explainability...")


# ================================================================
# 17.1 LIME EXPLAINER
# ================================================================

explainer = LimeTextExplainer(

    class_names=['Other', 'Risk']

)


# ================================================================
# 17.2 HEURISTIC PREDICTOR
# ================================================================

def predictor(texts):

    risk_words = [

       "fraud",
       "debt",
       "payment",
       "loan",
       "credit",
       "risk",
       "severity",
       "customer",
       "impact",
       "regulatory",
       "compliance",
       "recommendation",
       "charge",
       "identity",
       "unauthorized"
    ]

    probs = []

    for text in texts:

        text = text.lower()

        score = sum(

            word in text

            for word in risk_words

        ) / len(risk_words)

        probs.append([

            1 - score,

            score
        ])

    return np.array(probs)


# ================================================================
# 17.3 SELECT REPRESENTATIVE SAMPLE
# ================================================================

cot_rows = task1_df[

    task1_df['prompt_type'] == 'cot'
]

if len(cot_rows) > 0:

    sample_text = cot_rows[
    'output'
].dropna().astype(str).iloc[0][:1500]

else:

    sample_text = task1_df[
        'output'
    ].iloc[0]

print(
    "Selected text for LIME:",
    sample_text[:300]
)


# ================================================================
# 17.4 GENERATE EXPLANATION
# ================================================================

exp = explainer.explain_instance(

    sample_text,

    predictor,

    num_features=10
)

# ================================================================
# 17.5 FEATURE IMPORTANCE TABLE
# ================================================================

lime_features = exp.as_list()

lime_df = pd.DataFrame(

    lime_features,

    columns=['token', 'importance']
)

display(lime_df)

# ================================================================
# 18. SHAP EXPLAINABILITY
# ================================================================

print("\n===== SHAP EXPLAINABILITY =====\n")


# ================================================================
# 18.1 SHAP PREDICTOR
# ================================================================

def shap_predict(texts):

    risk_words = [
        "fraud",
        "debt",
        "payment",
        "loan",
        "credit"
    ]

    probs = []

    for t in texts:

        score = sum(
            w in t.lower()
            for w in risk_words
        ) / len(risk_words)

        probs.append(score)

    return np.array(probs)

# ================================================================
# 18.2 SHAP EXPLAINER
# ================================================================

explainer = shap.Explainer(

    shap_predict,

    shap.maskers.Text()
)


# ================================================================
# 18.3 SELECT REPRESENTATIVE OUTPUT
# ================================================================

cot_rows = task1_df[

    task1_df['prompt_type'] == 'cot'
]

if len(cot_rows) > 0:

    sample_text = cot_rows[
    'output'
].dropna().astype(str).iloc[0][:1500]

else:

    sample_text = task1_df[
        'output'
    ].iloc[0]

print(
    "Selected text for LIME:",
    sample_text[:300]
)


# ================================================================
# 18.4 GENERATE SHAP VALUES
# ================================================================

shap_values = explainer([sample_text])

# ================================================================
# 18.5 FEATURE IMPORTANCE TABLE
# ================================================================

shap_tokens = shap_values[0].data

shap_scores = shap_values[0].values


shap_df = pd.DataFrame({

    'token': shap_tokens,

    'importance': shap_scores
})


# ================================================================
# 18.6 ABSOLUTE IMPORTANCE
# ================================================================

shap_df['abs_importance'] = (

    shap_df['importance'].abs()
)

shap_df = shap_df.sort_values(

    by='abs_importance',

    ascending=False
)


display(shap_df.head(15))



# ================================================================
# 19. INTEGRATED GRADIENTS (GENERIC TRANSFORMER VERSION)
# ================================================================

import torch

from transformers import (
    AutoTokenizer,
    AutoModel
)

from captum.attr import IntegratedGradients


print("\n===== INTEGRATED GRADIENTS EXPLAINABILITY =====\n")

# ================================================================
# 19.1 LOAD GENERIC DISTILBERT
# ================================================================

tokenizer = AutoTokenizer.from_pretrained(
    "distilbert-base-uncased"
)

model = AutoModel.from_pretrained(
    "distilbert-base-uncased"
)

model.eval()


# ================================================================
# 19.2 REPRESENTATIVE GENERATED OUTPUT
# ================================================================

cot_rows = task1_df[

    task1_df['prompt_type'] == 'cot'
]

if len(cot_rows) > 0:

    sample_text = cot_rows[
    'output'
].dropna().astype(str).iloc[0][:1500]

else:

    sample_text = task1_df[
        'output'
    ].iloc[0]

print(
    "Selected text for LIME:",
    sample_text[:300]
)


# ================================================================
# 19.3 TOKENIZATION
# ================================================================

inputs = tokenizer(

    sample_text,

    return_tensors="pt",

    truncation=True,

    padding=True
)


# ================================================================
# 19.4 EMBEDDING LAYER
# ================================================================

embedding_layer = (
    model.embeddings.word_embeddings
)


# ================================================================
# 19.5 FORWARD FUNCTION
# ================================================================

def forward_func_emb(

    inputs_embeds,

    attention_mask
):

    outputs = model(

        inputs_embeds=inputs_embeds,

        attention_mask=attention_mask
    )

    # Mean pooled embedding score
    pooled_output = outputs.last_hidden_state.mean(dim=1)

    # Use mean activation as attribution target
    return pooled_output.mean(dim=1)


# ================================================================
# 19.6 INTEGRATED GRADIENTS
# ================================================================

ig = IntegratedGradients(
    forward_func_emb
)


# ================================================================
# 19.7 INPUT EMBEDDINGS
# ================================================================

input_embeddings = embedding_layer(
    inputs['input_ids']
)


# ================================================================
# 19.8 CLEAR GRADIENTS
# ================================================================

model.zero_grad()


# ================================================================
# 19.9 COMPUTE ATTRIBUTIONS
# ================================================================

attributions, delta = ig.attribute(

    inputs=input_embeddings,

    additional_forward_args=(

        inputs['attention_mask'],
    ),

    return_convergence_delta=True
)


# ================================================================
# 19.10 TOKENS + SCORES
# ================================================================

tokens = tokenizer.convert_ids_to_tokens(
    inputs['input_ids'][0]
)

scores = attributions.sum(

    dim=-1

).squeeze(0).detach().numpy()


# ================================================================
# 19.11 DATAFRAME
# ================================================================

ig_df = pd.DataFrame({

    'token': tokens,

    'importance': scores
})


# ================================================================
# 19.12 REMOVE SPECIAL TOKENS
# ================================================================

ig_df = ig_df[

    ~ig_df['token'].isin([

        '[CLS]',

        '[SEP]'
    ])
]

# ================================================================
# 19.13 REMOVE SYMBOLS / SHORT TOKENS
# ================================================================

ig_df = ig_df[

    ig_df['token'].str.len() > 2
]

ig_df = ig_df[

    ~ig_df['token'].str.contains(
        r'[^a-zA-Z]'
    )
]


# ================================================================
# 19.14 ABSOLUTE IMPORTANCE
# ================================================================

ig_df['abs_importance'] = (

    ig_df['importance'].abs()
)


# ================================================================
# 19.15 SORT
# ================================================================

ig_df = ig_df.sort_values(

    by='abs_importance',

    ascending=False
)


# ================================================================
# 19.16 DISPLAY
# ================================================================

display(ig_df.head(20))


# ================================================================
# 19.17 VISUALIZATION
# ================================================================

# ================================================================
# FINAL INTEGRATED GRADIENTS VISUALIZATION
# ================================================================

top_ig = ig_df.sort_values(
    by='abs_importance',
    ascending=False
).head(10)

ig_colors = plt.cm.Paired(

    np.linspace(
        0,
        1,
        len(top_ig)
    )
)

plt.figure(figsize=(10,5))

plt.bar(

    top_ig['token'],

    top_ig['abs_importance'],

    color=ig_colors

)

plt.xticks(rotation=45)

plt.title(
    "Task 1 - Integrated Gradients Token Importance"
)

plt.tight_layout()

plt.show()



# ================================================================
# 24. FINAL LIME VISUALIZATION
# ================================================================

lime_plot_df = lime_df.sort_values(

    by='importance',

    ascending=False
).head(10)

colors = plt.cm.tab20(

    np.linspace(
        0,
        1,
        len(lime_plot_df)
    )
)

plt.figure(figsize=(10,5))

plt.bar(

    lime_plot_df['token'],

    lime_plot_df['importance'],

    color=colors

)

plt.xticks(rotation=45)

plt.title(
    "Task 1 - LIME Token Importance"
)

plt.show()



# ================================================================
# 21. FINAL SHAP VISUALIZATION
# ================================================================

top_shap = shap_df.sort_values(

    by='abs_importance',

    ascending=False
).head(10)

plt.figure(figsize=(10,5))

colors = plt.cm.Set3(

    np.linspace(
        0,
        1,
        len(top_shap)
    )
)

plt.bar(

    top_shap['token'],

    top_shap['abs_importance'],

    color=colors

)

plt.xticks(rotation=45)

plt.title("SHAP Token Attribution")

plt.show()



# ================================================================
# 22. FINAL INTEGRATED GRADIENTS VISUALIZATION
# ================================================================

top_ig = ig_df.sort_values(

    by='abs_importance',

    ascending=False
).head(10)

plt.figure(figsize=(10,5))

colors = plt.cm.Paired(

    np.linspace(
        0,
        1,
        len(top_ig)
    )
)

plt.bar(

    top_ig['token'],

    top_ig['abs_importance'],

    color=colors

)

plt.xticks(rotation=45)

plt.title("Integrated Gradients Token Importance")

plt.show()


print("\n===== FINAL PERFORMANCE VISUALIZATIONS =====\n")

# ================================================================
# 23. FINAL PERFORMANCE VISUALIZATIONS
# ================================================================

import matplotlib.pyplot as plt
import seaborn as sns


# ================================================================
# 23.1 BERTScore Plot
# ================================================================

plt.figure(figsize=(10, 5))

prompt_palette = {

    'zero_shot': '#1f77b4',
    'few_shot': '#ff7f0e',
    'role_based': '#2ca02c',
    'instruction': '#d62728',
    'constraint': '#9467bd',
    'cot': '#8c564b'
}

sns.barplot(

    data=bert_summary,

    x='prompt_type',

    y='bert_f1',

    palette=prompt_palette
)

plt.title("TASK1:  Risk Analysis & Complaint Summarization - BERTScore Comparison")

plt.xticks(rotation=45)

plt.show()



# ================================================================
# 23.2 Hallucination Plot
# ================================================================

plt.figure(figsize=(10, 5))

sns.barplot(

    data=summary_task1,

    x='prompt_type',

    y='hallucination',

    palette=prompt_palette
)

plt.title("TASK1:  Risk Analysis & Complaint Summarization - Hallucination Comparison")

plt.xticks(rotation=45)

plt.show()



# ================================================================
# 23.3 Regulatory Score Plot
# ================================================================

plt.figure(figsize=(10, 5))

sns.barplot(

    data=summary_task1,

    x='prompt_type',

    y='regulatory_score',

    palette=prompt_palette
)

plt.title("TASK1:  Risk Analysis & Complaint Summarization - Regulatory Awareness")

plt.xticks(rotation=45)

plt.show()


plt.figure(figsize=(10, 5))

sns.barplot(

    data=summary_task1,

    x='prompt_type',

    y='completeness_score',

    palette=prompt_palette
)

plt.title(
    "TASK1 - Completeness Score Comparison"
)

plt.xticks(rotation=45)

plt.show()


# ================================================================
# 24. Export
# ================================================================

print("\n===== EXPORTING TABLES =====\n")

summary_task1.to_excel(

    f"{MODEL_SHORT}_task1_summary.xlsx",

    index=False
)

bert_summary.to_excel(

    f"{MODEL_SHORT}_task1_bertscore_summary.xlsx",

    index=False
)

final_task1_summary.to_excel(

    f"{MODEL_SHORT}_task1_final_summary.xlsx",

    index=False
)

lime_df.to_excel(

    f"{MODEL_SHORT}_task1_lime_importance.xlsx",

    index=False
)

shap_df.to_excel(

    f"{MODEL_SHORT}_task1_shap_importance.xlsx",

    index=False
)

ig_df.to_excel(

    f"{MODEL_SHORT}_task1_ig_importance.xlsx",

    index=False
)

print("Tables exported successfully")

plt.figure(figsize=(10,5))

sns.barplot(
    data=summary_task1,
    x='prompt_type',
    y='factual_score',
    palette=prompt_palette
)

plt.title(
    "Task1 - Factual Score Comparison"
)

plt.xticks(rotation=45)

plt.tight_layout()


FIG_DPI = 300

plt.savefig(

    f"{MODEL_SHORT}_task1_factual_score.png",

    dpi=FIG_DPI,

    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=summary_task1,
    x='prompt_type',
    y='hallucination',
    palette=prompt_palette
)

plt.title(
    "Task1 - Hallucination Comparison"
)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(

    f"{MODEL_SHORT}_task1_hallucination.png",

    dpi=FIG_DPI,

    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=summary_task1,
    x='prompt_type',
    y='regulatory_score',
    palette=prompt_palette
)

plt.title(
    "Task1 - Regulatory Score Comparison"
)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(

    f"{MODEL_SHORT}_task1_regulatory_score.png",

    dpi=FIG_DPI,

    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

sns.barplot(
    data=summary_task1,
    x='prompt_type',
    y='completeness_score',
    palette=prompt_palette
)

plt.title(
    "Task1 - Completeness Score Comparison"
)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(

    f"{MODEL_SHORT}_task1_completeness_score.png",

    dpi=FIG_DPI,

    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(12,6))

sns.heatmap(
    heatmap_data,
    annot=True,
    cmap='Blues'
)

plt.title(
    "Task1 Prompt Performance Heatmap"
)

plt.tight_layout()

plt.savefig(

    f"{MODEL_SHORT}_task1_heatmap.png",

    dpi=FIG_DPI,

    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    lime_plot_df['token'],
    lime_plot_df['importance'],
    color=colors
)

plt.xticks(rotation=45)

plt.title(
    "Task1 LIME Token Importance"
)

plt.tight_layout()

plt.savefig(

    f"{MODEL_SHORT}_task1_lime.png",

    dpi=FIG_DPI,

    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    top_shap['token'],
    top_shap['abs_importance'],
    color=colors
)

plt.xticks(rotation=45)

plt.title(
    "Task1 SHAP Token Importance"
)

plt.tight_layout()

plt.savefig(

    f"{MODEL_SHORT}_task1_shap.png",

    dpi=FIG_DPI,

    bbox_inches='tight'
)

plt.close()

plt.figure(figsize=(10,5))

plt.bar(
    top_ig['token'],
    top_ig['abs_importance'],
    color=colors
)

plt.xticks(rotation=45)

plt.title(
    "Task1 Integrated Gradients"
)

plt.tight_layout()

plt.savefig(

    f"{MODEL_SHORT}_task1_integrated_gradients.png",

    dpi=FIG_DPI,

    bbox_inches='tight'
)

plt.close()


print("\n===== TASK 1 EXECUTION COMPLETED =====\n")

print(
    "Total Outputs Generated:",
    len(task1_df)
)

print(
    "Total Prompt Types:",
    task1_df['prompt_type'].nunique()
)

print(
    "Model Used:",
    MODEL_NAME
)

print(
    "Total API Errors:",
    error_counter
)


```
